```python
import pickle
import cv2
import numpy as np
from matplotlib import pyplot as plt

# You need to point to the correct path where the pickles are stored
# i do ../ because my current working directory is Models, so we go up one level to main project folder : SAR-Denoising-Project, and then down to ETL/data/pickles
with open('../ETL/data/pickles/processed.pkl', 'rb') as file:
	data = pickle.load(file)
with open('../ETL/data/pickles/patches.pkl', 'rb') as file:
	patches = pickle.load(file)
```

```python
len(patches['bsd68']['clean']) + len(patches['set12']['clean'])
```

```python
# we have 3 datasets: bsd68, set12, div2k
# bsd68 and set12 are easy to access to, example :

i = 5
img = data['bsd68']['clean'][i]
noisy_img = data['bsd68']['noisy'][i]

# let's visualise the images with cv2, the images are in numpy array format np.float32 between 0 and 1
plt.figure(figsize=(10,5))
# Noisy
plt.subplot(1, 2, 1)
plt.imshow(noisy_img, cmap='gray')
plt.title("bsd68 Noisy : " + str(i + 1))
plt.axis("off")
# Clean
plt.subplot(1, 2, 2)
plt.imshow(img, cmap='gray')
plt.title("bsd68 Clean : " + str(i + 1))
plt.axis("off")

plt.tight_layout()
plt.show()
print("Image shape:", img.shape)
print("Noisy Image shape:", noisy_img.shape)
print("Image mean:", np.mean(img))
print("Noisy Image mean:", np.mean(noisy_img))
print("Image std:", np.std(img))
print("Noisy Image std:", np.std(noisy_img))
```
